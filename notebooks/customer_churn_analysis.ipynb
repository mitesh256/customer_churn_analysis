```python
# Total customers
total_customers = len(df)

# Churned and retained customers
churned_customers = (df["Churn"] == "Yes").sum()
retained_customers = (df["Churn"] == "No").sum()

# Calculate churn and retention rates
churn_rate = (churned_customers / total_customers) * 100
retention_rate = (retained_customers / total_customers) * 100

# Display summary
kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Retained Customers",
        "Churn Rate (%)",
        "Retention Rate (%)"
    ],
    "Value": [
        total_customers,
        churned_customers,
        retained_customers,
        round(churn_rate, 2),
        round(retention_rate, 2)
    ]
})

print(kpi_summary)
```

```python
contract_analysis = df.groupby("Contract").agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=(
        "Churn",
        lambda x: (x == "Yes").sum()
    )
)

contract_analysis["Churn_Rate_Percent"] = (
    contract_analysis["Churned_Customers"]
    / contract_analysis["Total_Customers"] * 100
).round(2)

display(
    contract_analysis.sort_values(
        "Churn_Rate_Percent",
        ascending=False
    )
)
```

```python

df["Tenure_Group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months",
            "25-48 Months", "49-72 Months"]
)

tenure_analysis = df.groupby(
    "Tenure_Group",
    observed=False
).agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=(
        "Churn",
        lambda x: (x == "Yes").sum()
    )
)

tenure_analysis["Churn_Rate_Percent"] = (
    tenure_analysis["Churned_Customers"]
    / tenure_analysis["Total_Customers"] * 100
).round(2)

display(tenure_analysis)
```

```python
highest_churn_contract = (
    contract_analysis["Churn_Rate_Percent"].idxmax()
)

highest_churn_contract_rate = (
    contract_analysis["Churn_Rate_Percent"].max()
)

highest_churn_tenure = (
    tenure_analysis["Churn_Rate_Percent"].idxmax()
)

highest_churn_tenure_rate = (
    tenure_analysis["Churn_Rate_Percent"].max()
)

print("CUSTOMER CHURN BUSINESS SUMMARY")
print("-" * 40)

print(f"Total customers: {total_customers}")
print(f"Churned customers: {churned_customers}")
print(f"Overall churn rate: {churn_rate:.2f}%")
print(f"Retention rate: {retention_rate:.2f}%")

print(
    f"\nHighest-churn contract type: "
    f"{highest_churn_contract} "
    f"({highest_churn_contract_rate:.2f}%)"
)

print(
    f"Highest-churn tenure group: "
    f"{highest_churn_tenure} "
    f"({highest_churn_tenure_rate:.2f}%)"
)
```

```python
# Create a business insights report

business_insights = [
    "CUSTOMER CHURN ANALYSIS — BUSINESS INSIGHTS",
    "=" * 50,
    "",
    f"Total Customers: {total_customers}",
    f"Churned Customers: {churned_customers}",
    f"Retained Customers: {retained_customers}",
    f"Overall Churn Rate: {churn_rate:.2f}%",
    f"Retention Rate: {retention_rate:.2f}%",
    "",
    "KEY FINDINGS",
    "-" * 30,
    (
        f"1. {highest_churn_contract} contract customers "
        f"have the highest observed churn rate "
        f"({highest_churn_contract_rate:.2f}%)."
    ),
    (
        f"2. The {highest_churn_tenure} tenure group "
        f"has the highest observed churn rate "
        f"({highest_churn_tenure_rate:.2f}%)."
    ),
    "",
    "BUSINESS RECOMMENDATIONS",
    "-" * 30,
    "1. Investigate onboarding and early customer experience.",
    "2. Evaluate retention offers for month-to-month customers.",
    "3. Monitor churn rates by tenure and contract type.",
    "4. Test retention strategies before assuming they work.",
    "",
    "Note: These findings show associations, not causation."
]

report_text = "\n".join(business_insights)

print(report_text)
```
